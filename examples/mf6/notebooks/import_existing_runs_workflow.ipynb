# Import Existing MF6 Runs

This notebook demonstrates the archive-discovery workflow for legacy MODFLOW 6 runs.
It scans an existing run directory, browses the archive with `RunExplorer`, opens one run lazily from native MF6 files, and then optionally imports those runs into a lightweight `ProjectCatalog`.


```python
from pathlib import Path

from simple_modflow import explore_runs, import_run_archive, summarize_discovered_runs
```

```python
archive_root = Path(r"C:\\Users\\lukem\\mf6")
catalog_root = Path(r"C:\\Users\\lukem\\Python\\Projects\\simple_modflow\\examples\\mf6\\artifacts\\imported_mf6_catalog")
```

```python
discovered = summarize_discovered_runs(archive_root)
discovered.sort_values("run_id").head(20)
```

```python
archive = explore_runs(archive_root)
archive.summary().sort_values("run_id").head(20)
```

```python
archive.families()
```

```python
catalog = import_run_archive(
    archive_root,
    catalog_root,
    project_name="mf6_archive_catalog",
    model_spec="legacy_archive",
    skip_existing=True,
)

catalog.run_summary().sort_values("run_id").head(20)
```

```python
run_a = "cumb_v6r"
run_b = "cumb_v6s_botm13"

run = archive.open(run_a, verbosity_level=1)
run.summary()
```
